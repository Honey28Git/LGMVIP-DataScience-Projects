```python
import pandas as pd
import numpy as np
import seaborn as sns 
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore", category=UserWarning, module="IPython")
```

```python
test = pd.read_csv("test.csv")
```

```python
members = pd.read_csv('members_eda.csv')
songs = pd.read_csv('songs_eda.csv')
```

```python
# Merge test with members and songs
test = test.merge(members, on='msno', how='left')
test = test.merge(songs, on='song_id', how='left')
```

```python
print("Nulls after fill:\n", test.isnull().sum())
print("Duplicates:", test.duplicated().sum())
```

```python
# Drop rows with completely missing member data
test = test[~test['registration_year'].isna()]

# Fill missing song-related values (same logic as train)
fill_unknown = ['artist_name', 'composer', 'lyricist', 'name', 'primary_genre']
for col in fill_unknown:
    test[col] = test[col].fillna('unknown')

test['language'] = test['language'].fillna(-1)
test['song_length'] = test['song_length'].fillna(test['song_length'].median())
test['song_length_sec'] = test['song_length_sec'].fillna(test['song_length_sec'].median())
test['isrc_year'] = test['isrc_year'].fillna(0)
test.fillna({
    'source_system_tab': 'unknown',
    'source_screen_name': 'unknown',
    'source_type': 'unknown',
    'genre_ids': 'unknown',
    'isrc': 'missing'
}, inplace=True)


# Final checks
print("Nulls after fill:\n", test.isnull().sum())
print("Duplicates:", test.duplicated().sum())
```

```python
test.to_csv('test_eda.csv', index=False)
```
